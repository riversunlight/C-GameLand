```python
import math
import random
```

```python
# Rastrigin Function
# たくさんの局所最適解を持つ
class Problem:
  A = 10
  def object_func(self, para):
    k = 0
    for x in para:
        k += self.A + (x*x - self.A * math.cos(2*math.pi*x))
    return k
```

```python
class GaSolver:
  generations = 1
  childs = 10
  problem = Problem()

  def crossover(self, x1, x2):
    x = [0 for i in range(len(x1))]
    print(x)
    for i in range(len(x1)):
      r = random.random()
      if r <= 0.5:
        x[i] = x1[i]
      else:
        x[i] = x2[i]
    print(x)
    return x


  def mutate(self, x):
    for i in range(len(x)):
      x[i] += random.random()
    return x

  def init_solutions(self):
    xs = []
    for i in range(2):
      x = []
      for j in range(2):
        x.append(random.random()*100 - 50)
      xs.append(x)
    return xs

  def solve(self):
    solutions = self.init_solutions()
    print(solutions)

    for i in range(self.generations):
      print(f"generation: {i}")
      child_solutions = []
      for j in range(self.childs):
        child = self.crossover(solutions[0], solutions[1])
        child_solutions.append([self.problem.object_func(child), child])

      child_solutions = sorted(child_solutions)
      print(child_solutions)
      solutions = []
      for j in range(2):
        solutions.append(child_solutions[j][1])

    return solutions[0]
```

```python
solver = GaSolver()
problem = Problem()
ans = solver.solve()
print("ans:", ans)
print("object_func", problem.object_func(ans) )
```
